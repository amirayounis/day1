```python
#---- chat bot   
from langchain.agents import create_agent
chatbot=create_agent(
    llm      # init_chat_model   /"gpt-4.1-mini"
)
```

```python
from langchain.messages import HumanMessage
resp=chatbot.invoke({
    "messages":[
        HumanMessage(content="what is the capital of moon") 
})
```

```python
from pprint import 
pprint(resp)
```

```python
ai_c=resp["messages"][-1].content
```

```python
from langchain.messages import AIMessage
resp=chatbot.invoke({
    "messages":[
        HumanMessage(content="what is the capital of moon"),
        AIMessage(content=ai_c) ,
        HumanMessage(content="no give me it as a scince fiction writer ") ]
})
```

```python
chatbot=create_agent(
    "gpt-4.1"
    ,checkpointer=InMemorySaver()      # ram /fast / non presistante / demos 
)
```

```python
#   image 
resp=chatbot.invoke({
    "messages":[
        HumanMessage(content=[
            {
                "type":"text",
                "text":"give me reciepes of the ingerdiants in the image"
            },
            {
               "type":"image",
                "url": "https://www.efsa.europa.eu/sites/default/files/styles/social_media_share/public/2021-03/food-ingredients-packaging.jpg"
            }
        ]),
       ]
},
{
    "configurable":{
        "thread_id":"10"
    }
})
```

```python
print(resp["messages"][-1].content)
```

```python
import base64
with open("imgs/food-ingredients-packaging.jpg", "rb") as f:
    image_data = f.read()
image_base64 = base64.b64encode(image_data).decode("utf-8")
resp=chatbot.invoke({
    "messages":[
        HumanMessage(content=[
        {"type": "text", "text": "Describe the content of this image."},
        {
            "type": "image",
            "base64": image_base64,
            "mime_type": "image/jpeg",
        },
    ])
    ]
},
{
    "configurable":{
        "thread_id":"11"
    }
})
```

```python
print(resp["messages"][-1].content)
```

```python
import sqlite3
connection=sqlite3.connect("data.sqlite",check_same_thread=False)
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.sqlite import SqliteSaver

chatbot=create_agent(
    llm 
    ,checkpointer=SqliteSaver(conn=connection)      # ram /fast / non presistante / demos 
    ,system_prompt="You are a helpful science fiction writer."
)
```

```python
from langchain.messages import HumanMessage
chatbot.invoke({
    "messages":[
        HumanMessage(content="what is the capital of moon") ]
},
{
    "configurable":{"thread_id":"33"}
}
)
```
